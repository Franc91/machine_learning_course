### <a name='0'></a> Import bibliotek

````python
import numpy as np
import pandas as pd
import sklearn

sklearn.__version__
````

### <a name='1'></a> Wygenerowanie danych

````python
data = {
    'size': ['XL', 'L', 'M', 'L', 'M'],
    'color': ['red', 'green', 'blue', 'green', 'red'],
    'gender': ['female', 'male', 'male', 'female', 'female'],
    'price': [199.0, 89.0, 99.0, 129.0, 79.0],
    'weight': [500, 450, 300, 380, 410],
    'bought': ['yes', 'no', 'yes', 'no', 'yes']
}

df_raw = pd.DataFrame(data=data)
df_raw
````

### <a name='2'></a> Utworzenie kopii danych

````python
df = df_raw.copy()
df.info()
````

### <a name='4'></a> LabelEncoder

````python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le.fit(df['bought'])
le.transform(df['bought'])
````

````python
le.fit_transform(df['bought'])
````

````python
le.classes_
````

````python
df['bought'] = le.fit_transform(df['bought'])
df
````

````python
le.inverse_transform(df['bought'])
````

````python
df['bought'] = le.inverse_transform(df['bought'])
df
````

### <a name='5'></a> OneHotEncoder

````python
from scipy import sparse
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)
encoder.fit_transform(df[['size']])
````

````python
encoder.categories_
````

````python
encoder = OneHotEncoder(drop='first', sparse_output=False)  #drop usuwa 1
encoder.fit_transform(df[['size']])
````

W wyniku działania tego kodu, kolumna 'size' zostanie zakodowana w taki sposób, że dla każdej unikalnej wartości w 'size' zostanie utworzona nowa kolumna binarna (0 lub 1), a jedna z tych kolumn zostanie pominięta (drop='first').

````python
df
````

### <a name='6'></a> Pandas *get_dummies()*

````python
pd.get_dummies(data=df, drop_first=True, prefix='new', prefix_sep='-').astype(int)
````

### <a name='7'></a> Standaryzacja - StandardScaler

std() - pandas nieobciążony  
std() - numpy obciążony

````python
print(f"{df['price']}\n")
print(f"Średnia: {df['price'].mean()}")
print(f"Odchylenie standardowe: {df['price'].std():.4f}")
````

````python
(df['price'] - df['price'].mean()) / df['price'].std()
````

````python
def standarize(x):
  return (x - x.mean()) / x.std()

standarize(df['price'])
````

````python
from sklearn.preprocessing import scale

scale(df['price'])
````

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit_transform(df[['price']])
````

````python
scaler_price = StandardScaler
df[['price', 'weight']] = scaler.fit_transform(df[['price', 'weight']])
df
````

### <a name='8'></a> Przygotowanie danych do modelu

````python
df = df_raw.copy()
df
````

````python
le = LabelEncoder()

df['bought'] = le.fit_transform(df['bought'])

scaler = StandardScaler()
df[['price', 'weight']] = scaler.fit_transform(df[['price', 'weight']])

df = pd.get_dummies(data = df, drop_first=True)
df
````
